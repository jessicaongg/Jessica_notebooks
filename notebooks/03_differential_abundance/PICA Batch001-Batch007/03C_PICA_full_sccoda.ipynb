### Full-Cohort Differential Abundance Analysis using scCODA
*Dataset: PICA Batch001–Batch007 (PICA0001–PICA0128)* \
*Analysis: Global compositional differential abundance*


### Background

scCODA is a compositional analysis method that models donor-level cell-type counts. It is used to identify changes in immune-cell abundance while accounting for differences in overall cell composition between donors.

#### Aim
This notebook performs full-cohort differential abundance analysis using scCODA.

The annotated PICA single-cell dataset is combined with donor-level metadata, then converted into sample-by-cell-type count data for compositional modelling. scCODA is used to test whether immune cell-type proportions are associated with age, sex and batch.

*Note: Previous scVI workflows were performed on lineage-specific datasets (CD4, CD8 and NK subsets). In contrast, scCODA was performed using the full annotated PICA dataset, with donor metadata re-integrated to preserve all immune populations required for global compositional modelling. As scCODA operates on donor-level cell-type counts, scVI latent embeddings were not required.*

## scCODA

### Prepare sccoda input
Convert cell-level annotations into donor-level Sample-by-Cell-Type Count Matrix required for scCODA analysis.

```python
# build sccoda input from cell-level data
def make_sccoda_from_lineage(
    adata,
    cell_type_col="pica_cell_type_complete_01_gdT_corrected",   
    sample_col="pica_id",       
    covariates=("Age_years","Sex", "Batch","Age"), 
):
    sccoda_data = sccoda_model.load(
        adata,
        type="cell_level",
        generate_sample_level=True,
        cell_type_identifier=cell_type_col,
        sample_identifier=sample_col,
        covariate_obs=list(covariates),
    )
    return sccoda_data
```

```python
# Aggregate annotated cells into donor-level cell-type counts
sccoda_full = make_sccoda_from_lineage(adata, cell_type_col="pica_cell_type_complete_01_gdT_corrected")
```

```python
sccoda_full
```

```python
# Extract sample-level cell-type count matrix
coda = sccoda_full["coda"].copy()
```

```python
# Format Covariates
coda.obs["Batch"] = coda.obs["Batch"].astype("category")
coda.obs["Age_years"] = coda.obs["Age_years"].astype("float")
coda.obs["Sex"] = coda.obs["Sex"].astype("category")
coda.obs["Age"] = coda.obs["Age"].astype("int")
```

```python
coda
```

```python
del sccoda_full
```

```python
# helper function to prepare sccoda model
def prepare_sccoda_age(sccoda_data, modality_key="coda", ref="automatic"):
    formula = "~ Age_years * Sex + Batch"

    sccoda_data = sccoda_model.prepare(
        sccoda_data,
        modality_key=modality_key,
        formula=formula,
        reference_cell_type=ref,  # or e.g. "CD4_naive"
    )
    return sccoda_data
```

```python
# Helper function to fit sccoda model
def run_sccoda(sccoda_data, modality_key="coda", rng_key=0):
    sccoda_model.run_nuts(
        sccoda_data,
        modality_key=modality_key,
        num_warmup=2000,
        num_samples=5000,
        rng_key=rng_key,
    )
    return sccoda_data
```

## Run sccoda

### Automatic ref cell type selection

```python
# Prepare scCODA model using automatic reference-cell selection
full_cell = prepare_sccoda_age(coda.copy(), ref="automatic")
```

```python
# Run scCODA model 
full_cell = run_sccoda(full_cell, rng_key=1)
```

## Donor-Level Cell-Type Proportion Analysis

Calculate and visualise donor-level cell-type proportions to aid interpretation of differential abundance results.

```python
# counts per sample x celltype
counts = pd.DataFrame(
    full_cell.X,
    index=full_cell.obs_names,
    columns=full_cell.var_names,
)

# convert to proportions
props = counts.div(counts.sum(axis=1), axis=0)

# metadata
meta = full_cell.obs.copy()

# long format for plotting
df = (
    props.join(meta[["Age_years", "Age"]])
    .reset_index(names="sample")
    .melt(
        id_vars=["sample", "Age_years", "Age"],
        var_name="pica_cell_type_complete_01_gdT_corrected",
        value_name="Proportion",
    )
)
```

```python
# Group cells into major lineages
lineage_map = {
    "CD4 Tem": "CD4 T cells",
    "CD4 cytoT": "CD4 T cells",
    "CD4 naive/Tcm": "CD4 T cells",
    "Tfh": "CD4 T cells",
    "Treg": "CD4 T cells",
    "CD8 Tem": "CD8 T cells",
    "CD8 naive/Tcm": "CD8 T cells",
    "CD8 GZMK+ naive/Tcm": "CD8 T cells",
    "DN T": "T cells",
    "DP T": "T cells",
    "gdT": "T cells",

    "CD56_bright CD16- NK": "NK",
    "CD56_dim CD16+ KLRC2+ NK": "NK",
    "CD56_dim CD16+ NK": "NK",
    "ILC1": "ILC",
    "ILC2": "ILC",
    "nILC2": "ILC",

    "Naive B cell": "B cells",
    "Non-switched memory B cell": "B cells",
    "Switched memory B cell": "B cells",
    "Atypical memory B cell": "B cells",
    "Transitional B cell": "B cells",
    "Plasma cell": "B cells",

    "Classical monocyte": "Myeloid",
    "Intermediate monocyte": "Myeloid",
    "CX3CR1_high ncMonocyte": "Myeloid",
    "CX3CR1_low ncMonocyte": "Myeloid",
    "cDC1": "Myeloid",
    "cDC2": "Myeloid",
    "pDC": "Myeloid",

    "MAIT": "Other",
    "NKT": "Other",
    "Platelet": "Other",
    "HSPC": "Other",
    "Proliferating lymphocyte": "Other",
    "CR2- B cell": "Other",
    "GNLY+ B": "Other"

}
```

```python
# Define lineage and cell-type order
df["lineage"] = df["pica_cell_type_complete_01_gdT_corrected"].map(lineage_map)

# fill anything unmapped
df["lineage"] = df["lineage"].fillna("Other")

# desired lineage order
lineage_order = [
    "CD4 T cells",
    "CD8 T cells",
    "T cells",
    "NK",
    "ILC",
    "B cells",
    "Myeloid",
    "Other"
]

# desired cell-type order within each lineage
cell_type_order = [
    "CD4 naive/Tcm", "CD4 Tem", "CD4 cytoT", "Tfh", "Treg",
    "CD8 naive/Tcm", "CD8 GZMK+ naive/Tcm", "CD8 Tem",
    "DN T", "DP T", "gdT",
    "CD56_bright CD16- NK", "CD56_dim CD16+ KLRC2+ NK", "CD56_dim CD16+ NK",
    "ILC1", "ILC2", "nILC2",
    "Naive B cell", "Transitional B cell", "Atypical memory B cell",
    "Non-switched memory B cell", "Switched memory B cell", "Plasma cell",
    "Classical monocyte", "Intermediate monocyte",
    "CX3CR1_high ncMonocyte", "CX3CR1_low ncMonocyte",
    "cDC1", "cDC2", "pDC",
    "MAIT", "NKT", "GNLY+ B", "HSPC", "Platelet", "Proliferating lymphocyte", "CR2- B cell"
]

# apply categorical ordering
df["lineage"] = pd.Categorical(df["lineage"], categories=lineage_order, ordered=True)
df["pica_cell_type_complete_01_gdT_corrected"] = pd.Categorical(
    df["pica_cell_type_complete_01_gdT_corrected"],
    categories=cell_type_order,
    ordered=True
)

# sort once
df = df.sort_values(["lineage", "pica_cell_type_complete_01_gdT_corrected"])
```

```python
# Visualise Cell-Type Proportions Across Age
for lineage in lineage_order:
    sub_df = df[df["lineage"] == lineage].copy()
    
    if sub_df.empty:
        continue
    
    cell_types = [x for x in cell_type_order if x in sub_df["pica_cell_type_complete_01_gdT_corrected"].astype(str).unique()]
    
    n_cols = 4
    n_rows = math.ceil(len(cell_types) / n_cols)

    fig, axes = plt.subplots(n_rows, n_cols, figsize=(5 * n_cols, 3.5 * n_rows))
    axes = np.array(axes).flatten()

    for ax, cell in zip(axes, cell_types):
        tmp = sub_df[sub_df["pica_cell_type_complete_01_gdT_corrected"].astype(str) == cell]
        tmp.boxplot(column="Proportion", by="Age", grid=False, ax=ax)

        ax.set_xticks(range(1, 20))
        ax.set_xticklabels(range(0, 19), fontsize=8)
        ax.set_xlim(0.5, 19.5)
        ax.set_title(cell, fontsize=10)
        ax.set_xlabel("")
        ax.set_ylabel("")

    for ax in axes[len(cell_types):]:
        ax.axis("off")

    plt.suptitle("")
    fig.suptitle(f"{lineage}: proportions by age", fontsize=14)
    fig.supxlabel("Age", fontsize=11)
    fig.supylabel("Proportion of cells", fontsize=11)
    plt.tight_layout(rect=[0.04, 0.04, 1, 0.95])
    plt.show()
```

```python
# barplot of mean composition by age
props_age = (props.join(meta["Age"]).groupby("Age").mean().reindex(range(0, 19)))

ax = props_age.plot(kind="bar", stacked=True, figsize=(10, 4))
ax.set_xlabel("Age")
ax.set_ylabel("Mean proportion")
ax.set_title("Mean composition by Age")
ax.legend(bbox_to_anchor=(1.02, 1), loc="upper left", title="Cell type")
plt.tight_layout()
plt.show()
```

```python
# Plot mean composition by age separately for each broad lineage
props_age = props.join(meta["Age"]).groupby("Age").mean().reindex(range(0, 19))

for lineage in lineage_order:
    lineage_cells = [ct for ct, grp in lineage_map.items() if grp == lineage]
    lineage_cells = [ct for ct in cell_type_order if ct in lineage_cells and ct in props_age.columns]
    
    if len(lineage_cells) == 0:
        continue
    
    ax = props_age[lineage_cells].plot(
        kind="bar",
        stacked=True,
        figsize=(10, 4)
    )
    ax.set_xlabel("Age")
    ax.set_ylabel("Mean proportion")
    ax.set_title(f"{lineage}: mean composition by age")
    ax.legend(bbox_to_anchor=(1.02, 1), loc="upper left", title="Cell type")
    plt.tight_layout()
    plt.show()
```

## Visualisation of scCODA Results
Summarise differential abundance effects across immune populations using parameter estimates and inclusion probabilities.

### Prepare Results for Visualisation

Filter and organise scCODA results to facilitate comparison of effect estimates across selected immune populations.

```python
# Remove batch covariates from plotting to focus on age, sex and age-by-sex effects
effects_full_plot = effects_full[
    ~effects_full.index.get_level_values("Covariate").str.startswith("Batch")
]
```

```python
# Convert scCODA effects table into plotting format
effects_full_plot = effects_full_plot.reset_index()

# Add broad lineage information to each cell type
effects_full_plot["lineage"] = effects_full_plot["Cell Type"].map(lineage_map).fillna("Other")

# Apply lineage ordering
effects_full_plot["lineage"] = pd.Categorical(
    effects_full_plot["lineage"],
    categories=lineage_order,
    ordered=True
)

# Apply cell-type ordering
effects_full_plot["Cell Type"] = pd.Categorical(
    effects_full_plot["Cell Type"],
    categories=cell_type_order,
    ordered=True
)

# Sort and remove duplicate / missing entries for plotting
effects_full_plot = effects_full_plot.sort_values(["lineage", "Cell Type"])
effects_full_plot = effects_full_plot.drop_duplicates()
effects_full_plot = effects_full_plot.dropna(subset=["Cell Type"]).copy()
```

### Effect Size Visualisation

Plot final parameter estimates for selected immune populations to assess the direction and magnitude of covariate effects.

```python
# Select key immune populations for visualising scCODA effects
selected_cells = [
    "CD4 naive/Tcm",
    "CD4 Tem",
    "Tfh",
    "CD4 cytoT",
    "Treg",
    "CD8 naive/Tcm",
    "CD8 GZMK+ naive/Tcm",
    "CD8 Tem",
    "CD56_bright CD16- NK",
    "CD56_dim CD16+ NK",
    "CD56_dim CD16+ KLRC2+ NK"
]

# Keep only selected cell types for plotting
plot_df = effects_full_plot[
    effects_full_plot["Cell Type"].isin(selected_cells)
].copy()

# Apply selected plotting order
plot_df["Cell Type"] = pd.Categorical(
    plot_df["Cell Type"],
    categories=selected_cells,
    ordered=True
)

plot_df = plot_df.sort_values("Cell Type")

# Plot final parameter estimates as a horizontal barplot
plt.figure(figsize=(8, 4))

sns.barplot(
    data=plot_df,
    y="Cell Type",
    x="Final Parameter",
    hue="Covariate",
    orient="h",
    order=selected_cells
)

plt.axvline(0, color="black", linewidth=1)
plt.xlabel("Final parameter (log-fold change)")
plt.ylabel("Cell type")
plt.title("scCODA effects by covariate")
plt.legend(title="Covariate", bbox_to_anchor=(1.02, 1), loc="upper left")

plt.tight_layout()
plt.show()
```

```python
# Plot individual final parameter estimates for selected cell populations
plt.figure(figsize=(9, 7))
sns.stripplot(
    data=plot_df,
    y="Cell Type",
    x="Final Parameter",
    hue="Covariate",
    dodge=True,
    orient="h",
    size=4
)
plt.axvline(0, color="black", linewidth=1)
plt.title("scCODA effects")
plt.tight_layout()
plt.show()
```

### Heatmap Visualization
- *Differential abundance effects*: summarise final parameter estimates across selected immune populations and covariates
- *Inclusion probabilities*: visualise inclusion probabilities to assess the support for each differential abundance effect.

```python
# Identify lineage breaks for separating cell groups in heatmaps
ordered_present = [ct for ct in cell_type_order if ct in beta_auto.index]

lineage_breaks = []
last_group = None
for i, ct in enumerate(ordered_present):
    grp = lineage_map.get(ct, "Other")
    if last_group is not None and grp != last_group:
        lineage_breaks.append(i)
    last_group = grp
```

```python
# Create heatmap matrix of final parameter estimates
beta_auto = (
    plot_df
    .pivot(index="Cell Type", columns="Covariate", values="Final Parameter")
    .reindex(selected_cells)
    .dropna(how="all")
)

# Plot heatmap of final parameter estimates
plt.figure(figsize=(7, 12))
ax = sns.heatmap(beta_auto, cmap="coolwarm", center=0)

for b in lineage_breaks:
    ax.hlines(b, *ax.get_xlim(), colors="black", linewidth=0.8)

plt.title("scCODA effects (Final Parameter)")
plt.xlabel("Covariate")
plt.xticks(rotation=45, ha="right")
plt.ylabel("Cell type")
plt.tight_layout()
plt.show()
```

```python
# Create heatmap matrix of inclusion probabilities
ip_auto = (
    plot_df
    .pivot(index="Cell Type", columns="Covariate", values="Inclusion probability")
    .reindex(selected_cells)
    .dropna(how="all")
)

# Plot inclusion probability heatmap
plt.figure(figsize=(7, 12))
ax = sns.heatmap(ip_auto, cmap="coolwarm", vmin=0, vmax=1)

for b in lineage_breaks:
    ax.hlines(b, *ax.get_xlim(), colors="black", linewidth=0.8)

plt.title("scCODA inclusion probabilities")
plt.xlabel("Covariate")
plt.xticks(rotation=45, ha="right")
plt.ylabel("Cell type")
plt.tight_layout()
plt.show()
```
