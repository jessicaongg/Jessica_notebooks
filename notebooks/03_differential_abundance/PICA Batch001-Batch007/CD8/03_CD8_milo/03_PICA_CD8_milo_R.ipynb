### Export CD8 Milo Object for MiloR Age-by-Sex Interaction Analysis
*Dataset: PICA Batch001–Batch007 (PICA0001–PICA0128)* \
*Analysis: Data export for age-by-sex interaction modelling in R*

#### Aim
This notebook prepares the CD8 Milo object for downstream age-by-sex interaction analysis in R.

The CD8 MuData object is converted into a simplified AnnData object containing the required expression matrix, metadata and embeddings. This exported object is then used for MiloR analysis in R to test whether age-associated CD8 neighbourhood abundance differs between male and female donors.

```python
import os
import scanpy as sc
import mudata as mu
import pandas as pd
import anndata as ad

outdir = "/scratch/user/s4575250/BIOX7014_Thesis/write/03_milo_DA/PICA_Batch001-Batch007"
os.makedirs(outdir, exist_ok=True)
```

```python
mdata_cd8 = mu.read("/scratch/user/s4575250/BIOX7014_Thesis/write/03_milo_DA/PICA_Batch001-Batch007/PICA_Batch001-Batch007_cd8_milo_nhoods_built.h5mu")
mdata_cd8
```

```python
# Extract the RNA modality from the CD8 MuData object for export
adata_cd8 = mdata_cd8["rna"].copy()
print(adata_cd8.obs.columns.tolist())
print(list(adata_cd8.obsm.keys()))

# Remove original MuData object to reduce memory usage
del mdata_cd8
```

```python
# Define metadata columns required for downstream MiloR age-by-sex interaction modelling
required_cols = ["pica_id", "Age_years", "Sex", "Batch", "pica_cell_type_complete_01_gdT_corrected"]

# Keep only cells with complete metadata for donor ID, age, sex, batch and annotation
keep = (
    adata_cd8.obs["pica_id"].notna()
    & adata_cd8.obs["Age_years"].notna()
    & adata_cd8.obs["Sex"].notna()
    & adata_cd8.obs["Batch"].notna()
    & adata_cd8.obs["pica_cell_type_complete_01_gdT_corrected"].notna()
)
adata_cd8 = adata_cd8[keep].copy()
```

```python
# Standardise metadata column types for downstream compatibility with R/MiloR
adata_cd8.obs["pica_id"] = adata_cd8.obs["pica_id"].astype(str)
adata_cd8.obs["ageyrs"] = pd.to_numeric(adata_cd8.obs["Age_years"], errors="coerce")
adata_cd8.obs["sex"] = adata_cd8.obs["Sex"].astype(str)
adata_cd8.obs["batch"] = adata_cd8.obs["Batch"].astype(str)
adata_cd8.obs["pica_cell_type_complete_01_gdT_corrected"] = adata_cd8.obs["pica_cell_type_complete_01_gdT_corrected"].astype(str)

adata_cd8 = adata_cd8[adata_cd8.obs["ageyrs"].notna()].copy()
```

```python
# Retain only metadata required for MiloR modelling and interpretation
obs_keep = ["pica_id", "ageyrs", "sex", "batch", "pica_cell_type_complete_01_gdT_corrected"]
# Retain gene metadata
var_df = adata_cd8.var.copy()

# Create simplified AnnData object for export
clean_cd8 = ad.AnnData(
    X=adata_cd8.X.copy(),
    obs=adata_cd8.obs[obs_keep].copy(),
    var=var_df.copy()
)

# Preserve scVI and UMAP embeddings for downstream neighbourhood construction and visualisation
clean_cd8.obsm["X_scVI"] = adata_cd8.obsm["X_scVI"].copy()
clean_cd8.obsm["X_umap"] = adata_cd8.obsm["X_umap"].copy()


```

```python
# Remove additional object slots that are not required for MiloR export
# This reduces file size and avoids cross-language I/O issues
clean_cd8.uns = {}
clean_cd8.layers = {}
clean_cd8.obsp = {}
clean_cd8.varm = {}
clean_cd8.varp = {}
```
